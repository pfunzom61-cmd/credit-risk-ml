Data profiling Step 3 : Identifying null counts

```python
import pandas as pd
from sqlalchemy import create_engine

# Connect to database

engine = create_engine("postgresql://postgres:PfunzoPostgres@localhost:5432/credit_risk")

# Pull the full table into a dataframe

print("Loading staging table...")
df = pd.read_sql("SELECT * FROM staging.application_train", engine)
print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns")

# NULL ANALYSIS 
# For every column, count nulls and express as a percentage


null_counts = df.isnull().sum()          # count nulls per column
null_pct = (null_counts / len(df)) * 100 # convert to percentage

# Combine into one clean summary table
null_summary = pd.DataFrame({
    'null_count': null_counts,
    'null_pct': null_pct.round(2)
})

# Sort by worst offenders first — these need decisions made about them
null_summary = null_summary[null_summary['null_count'] > 0].sort_values(
    'null_pct', ascending=False
)

print(f"\nColumns with nulls: {len(null_summary)} out of {df.shape[1]}")
print(null_summary.to_string())
```
